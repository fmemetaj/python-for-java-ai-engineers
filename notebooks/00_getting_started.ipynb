## 1. Start with a problem you already understand

Imagine a helpdesk system with two incoming tickets. In Java we might prototype with:

```java
List<Map<String, Object>> tickets = List.of(
    Map.of("id", 101, "subject", "Cannot sign in"),
    Map.of("id", 102, "subject", "Missing invoice")
);
System.out.println(tickets.get(0).get("subject"));
```

You don't need a lesson on collections. **The Python differences:** `[]` creates a list; `{}` with key-value pairs creates a dictionary; `tickets[0]` is the first item. Python can add type hints, but it doesn't require types on every variable.

````python
tickets = [
    {"id": 101, "subject": "Cannot sign in"},
    {"id": 102, "subject": "Missing invoice"},
]
print(tickets[0]["subject"])
assert tickets[0]["id"] == 101
````

## 2. Pause and predict

1. Does `tickets[1]` access the first ticket or the second?
2. Will changing a dictionary update the list that contains it?
3. If you edit the code without running it, what does the **kernel** remember?

**Your turn:** change the subject below to something different and run the cell twice.

````python
tickets[0]["subject"] = "Password reset fails"
print(tickets[0]["subject"])
print("Ticket count:", len(tickets))
assert len(tickets) == 2
````

### One Jupyter gotcha worth learning immediately

A notebook has a **live kernel**. If you run the second cell before the first, you may hit `NameError` because `tickets` hasn't been defined in that session. Use **Restart Kernel → Run All** when execution order gets confusing.
