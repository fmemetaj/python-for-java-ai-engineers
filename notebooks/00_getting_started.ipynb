## Familiar concepts, new syntax

Java's `List<Map<String, Object>>` is usually `list[dict[str, object]]` in Python. Dictionaries make quick prototypes convenient; Pydantic models add validation when the structure matters.

```python
tickets = [{"id": 1, "text": "I cannot sign in"}, {"id": 2, "text": "Where is my invoice?"}]
print(type(tickets).__name__, type(tickets[0]).__name__)
print(tickets[0]["text"])
```
