## A tiny runnable readiness check

Predict the output first. The code combines a Python comprehension, a typed request, an awaited workflow and simulated approval.

**Your turn:** alter the ticket's subject (but keep its body valid), then explain why the classification does or doesn't change.

```python
from mini_agent import Ticket, prepare_reply, simulate_send

categories = {name: len(name) for name in ["access", "billing"]}
assert categories == {"access": 6, "billing": 7}

ticket = Ticket(id=20, subject="Invoice missing", body="I need a copy of my invoice")
proposal = await prepare_reply(ticket)
print("Category:", proposal.category)
print("Approval simulation:", simulate_send(proposal, approved=False))
assert proposal.category == "billing"
assert simulate_send(proposal, approved=False)["would_send"] is False
```

## Answer key

1. Type annotations describe an intended contract; **Pydantic validates input at runtime**. Correct structure doesn't imply correct facts.
2. Sequential awaits wait one after another; `gather` schedules independent I/O **concurrently** and returns results in argument order.
3. `yield` emits data lazily; an **async generator** can await I/O between chunks.
4. A decorator is **executed at function definition time** and can wrap or register the callable; it is not merely metadata.
5. **No.** Real write actions need separate authorization, validation, audit, safeguards and approval where appropriate.
