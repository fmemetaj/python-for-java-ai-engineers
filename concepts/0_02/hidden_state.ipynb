Read each markdown cell **before** running the next code cell. The kernel is one long-lived process. **Run All** is the only order this file guarantees in CI.

```python
tickets = [{"id": 101, "subject": "Cannot sign in"}]
print(tickets[0]["subject"])
assert tickets[0]["id"] == 101
```

The next cell **reuses** `tickets`. If you ran only this second cell after a kernel restart, you would get `NameError`.

```python
tickets.append({"id": 102, "subject": "Missing invoice"})
print(len(tickets))
assert len(tickets) == 2
```

Rebinding the name `tickets` does not require a type declaration. The kernel now points the name at a new list. Old outputs above this cell may still show length 2 until you re-run them.

```python
tickets = [{"id": 201, "subject": "Password reset"}]
print(tickets[0]["id"])
assert len(tickets) == 1
```
